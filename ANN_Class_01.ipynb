```python
class Perceptron:
    def __init__(self,weight, bias):
        self.weight=np.array(weight)
        self.bias = np.array(bias)
    def step_function(self,x):
        return np.where(x>=0,1,0)
    def forward (self,input):
        net_input=np.dot(input,self.weight)+self.bias
        final = self.step_function(net_input)
        return final
```

```python
class MLP:
    def __init__(self):
        # creating instance of Perceptron class
        self.hidden=Perceptron(weight=[[1,-1],[1,-1]],
                              bias=[-0.5,1.5])
        self.output = Perceptron(weight=[1,1],
                                bias=-1.5)
    def predict(self,x):
            h_out =self.hidden.forward(x)
            final_out = self.output.forward (h_out)
            return final_out, h_out
```

```python
model = MLP()
data=[[0,1],[1,0],[1,1],[0,0]]
print(f"{'Input':<10} | {'Hidden Layer':<15} | {'Final Output'}")
print("-" * 45)
for point in data:
    pred, hidden_state = model.predict(point)
    print(f"{str(point):<10} | {str(hidden_state):<15} | {pred}")
```

```python
# parent class
class BaseNeuron:
    def __init__(self, weights,biases):
        self.weights=np.array(weights)
        self.biases=np.array(biases)
    def calculation (self,x):
        calc=np.dot(x,self.weights)+self.biases
        return calc
class StepNeuron (BaseNeuron):
    def forward (self,x):
        net=self.calculation(x)
        return np.where(net>=0,1,0)
class SigmoidNeuron (BaseNeuron):
    def forward(self, x):
        net = self.calculate_net(x)
        return 1 / (1 + np.exp(-net))
```

```python
# the composite class
class SimpleMLP:
    def __init__(self,w1,b1,w2,b2):
        self.layer1=StepNeuron(w1,b1) #hidden layer
        self.layer2=StepNeuron(w2,b2) #out layer
    def run(self,x):
        h_layer1=self.layer1.forward(x)
        out_layer2=self.layer2.forward(h_layer1)
        return h_layer1,out_layer2
```

```python
w_hidden=[[1,-1],[1,-1]]
b_hidden=[-0.5,1.5]
w_out=[1,1]
b_out=[-1.5]
```

```python
model = SimpleMLP(w_hidden,b_hidden,w_out,b_out)
```

```python
data=[[0,1],[1,0],[1,1],[0,0]]

print(f"{'Input':<10} | {'Hidden Layer':<15} | {'Final Output'}")
print("-" * 45)
for point in data:
    hidden_state, pred = model.run(point)
    print(f"{str(point):<10} | {str(hidden_state):<15} | {pred[0]}")
```

## Neurone

```python
import numpy as np

class NeuralNetwork:
    def __init__(self, input_size, hidden_size, output_size):
        self.input_size = input_size
        self.hidden_size = hidden_size
        self.output_size = output_size

        self.weights_input_hidden = np.random.randn(
            self.input_size, self.hidden_size)
        self.weights_hidden_output = np.random.randn(
            self.hidden_size, self.output_size)

        self.bias_hidden = np.zeros((1, self.hidden_size))
        self.bias_output = np.zeros((1, self.output_size))

    def sigmoid(self, x):
        return 1 / (1 + np.exp(-x))

    def sigmoid_derivative(self, x):
        return x * (1 - x)
        
    def feedforward(self, X):
        self.hidden_activation = np.dot(
            X, self.weights_input_hidden) + self.bias_hidden
        self.hidden_output = self.sigmoid(self.hidden_activation)
        
        self.output_activation = np.dot(
            self.hidden_output, self.weights_hidden_output) + self.bias_output
        self.predicted_output = self.sigmoid(self.output_activation)
        
        return self.predicted_output
    
    def backward(self, X, y, learning_rate):
        output_error = y - self.predicted_output
        output_delta = output_error * \
            self.sigmoid_derivative(self.predicted_output)
        
        hidden_error = np.dot(output_delta, self.weights_hidden_output.T)
        hidden_delta = hidden_error * self.sigmoid_derivative(self.hidden_output)
        
        self.weights_hidden_output += np.dot(self.hidden_output.T,
                                             output_delta) * learning_rate
        self.bias_output += np.sum(output_delta, axis=0,
                                   keepdims=True) * learning_rate
        self.weights_input_hidden += np.dot(X.T, hidden_delta) * learning_rate
        self.bias_hidden += np.sum(hidden_delta, axis=0,
                                   keepdims=True) * learning_rate

    def train(self, X, y, epochs, learning_rate):
        for epoch in range(epochs):
            output = self.feedforward(X)
            self.backward(X, y, learning_rate)
            if epoch % 4000 == 0:
                loss = np.mean(np.square(y - output))
                print(f"Epoch {epoch}, Loss:{loss}")
```

```python
X = np.array([[0, 0], [0, 1], [1, 0], [1, 1]])
y = np.array([[0], [1], [1], [0]])

nn = NeuralNetwork(input_size=2, hidden_size=4, output_size=1)
nn.train(X, y, epochs=10000, learning_rate=0.1)

output = nn.feedforward(X)
print("Predictions after training:")
print(output)
```

# Sequential Method

```python
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
from tensorflow.keras.optimizers import Adam
print ("tensorflow keras imported ...")
```

```python
from sklearn.datasets import load_iris
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
print ("sklearn imported ...")

# Load the dataset
iris = load_iris()

# Access the features (X) and the targets (y)
X = iris.data
y = iris.target
print(f"Features shape: {X.shape}") # (150, 4)
print(f"Target names: {iris.target_names}") # ['setosa' 'versicolor' 'virginica']
```

```python
# scaling X, y
X_train, X_test,y_train,y_test=train_test_split(X,y,test_size=0.20,random_state=42)
# scaling X_train, X_test
scaler = StandardScaler()
X_train_scaled =scaler.fit_transform(X_train)
X_test_scaled =scaler.transform(X_test)
# scaling y_train, y_test
y_train_encoded = tf.keras.utils.to_categorical(y_train) # OneHotEncoder
y_test_encoded = tf.keras.utils.to_categorical(y_test)
```

```python
model = Sequential()
model.add(Dense(10,activation='relu',input_shape=(X_train_scaled.shape[1],)))
model.add(Dense(10,activation='relu'))
model.add(Dense(3,activation='softmax'))
model
```

```python
model.summary()
```

```python
model.compile(loss='categorical_crossentropy',metrics=['accuracy'],optimizer=Adam(learning_rate=0.01))
```

```python
history = model.fit(X_train_scaled,y_train_encoded,epochs=100, batch_size=32, validation_split=0.2, verbose=1)
```

```python
print(history.history.keys())
```

```python
# Visualization of accuracy and validation accuracy
import matplotlib.pyplot as plt
plt.plot(history.history['accuracy'],label="training accuracy")
plt.plot(history.history['val_accuracy'],label="validation accuracy")
```

```python
# Evaluation
test_loss,test_accuracy = model.evaluate(X_test_scaled,y_test_encoded,verbose=0)
print (f" test loss: {test_loss:.2f}")
print (f"test accuracy: {test_accuracy:4f}")
```

```python
# prediction
import numpy as np
y_predict = model.predict(X_test_scaled)
class_predict = np.argmax(y_predict,axis=1)
print (class_predict)
```

```python
# Metrics
import pandas as pd
from sklearn.metrics import confusion_matrix, classification_report
cm = confusion_matrix(y_test, class_predict)
cm_df = pd.DataFrame(cm, index=iris.target_names, columns=iris.target_names)
cm_df.index.name = 'True'
cm_df.columns.name = "Prediction"
print ("confusion matrix")
print (cm_df)
print ('\n')
cr = classification_report(y_test,class_predict, target_names=iris.target_names)
print (cr)
```
